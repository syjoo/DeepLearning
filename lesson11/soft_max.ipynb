```python
def code(name, value):
    print(f"*** {name} ***\n{value}")
```

```python
def softmax(x, axis=-1):
    """수치 안정성을 위해 최댓값을 빼고 계산한다 (exp 오버플로 방지)."""
    x = x - x.max(axis=axis, keepdims=True)
    e = np.exp(x)
    
    return e / e.sum(axis=axis, keepdims=True)

x = np.array([
    [1, 5, 3],
    [4, 2, 8]
])

s = softmax(x)

code('softmax', s)
```

```python
axis = 0
x = np.array([2, 1, 0])
print(f"max({x}, axis={axis}) = {x.max(axis=axis, keepdims=True)}")
x = x - x.max(axis=axis, keepdims=True)
print(f"x - max({x}, axis={axis}) = {x}")
e = np.exp(x)
print(e.sum(axis=axis, keepdims=True))
print(e / e.sum(axis=axis, keepdims=True))
```
